# MwC 02 — Recompute 140-region MwC

This notebook is one ordered step in the analysis workflow.

```python
from pathlib import Path
import sys
import numpy as np
import pandas as pd
from IPython.display import Image, display

ROOT = Path.cwd()
if not (ROOT / 'analysis').is_dir():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'analysis'))
pd.set_option('display.max_columns', 30)
pd.set_option('display.max_rows', 100)
```

```python
from core import load_data
data = load_data()
meta = data['meta'].copy()
meta['finite_MwC_nodes'] = np.isfinite(data['mwc']).sum(axis=1)
display(meta.groupby(['analysis_cohort','analysis_arm','session'], dropna=False)
        .agg(sessions=('participant_id','size'), candidate=('mwc_candidate','sum'),
             finite_nodes_min=('finite_MwC_nodes','min'), finite_nodes_max=('finite_MwC_nodes','max')).reset_index())
```
