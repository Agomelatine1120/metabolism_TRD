# Main 05 — Longitudinal paired tests

This notebook is one ordered step in the analysis workflow.

```python
from pathlib import Path
import sys
import numpy as np
import pandas as pd
from IPython.display import Image, display

ROOT = Path.cwd()
if not (ROOT / 'analysis').is_dir():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'analysis'))
pd.set_option('display.max_columns', 30)
pd.set_option('display.max_rows', 100)
```

```python
from core import load_data, longitudinal_results
tests = longitudinal_results(load_data())
counts = (tests.groupby(['analysis_arm','modality'])
          .agg(n_tests=('feature','size'), paired_p_lt_0p05=('paired_p_lt_0p05','sum'),
               paired_p_lt_0p01=('paired_p_lt_0p01','sum')).reset_index())
display(counts)
display(tests.loc[tests.paired_p_lt_0p05])
```
