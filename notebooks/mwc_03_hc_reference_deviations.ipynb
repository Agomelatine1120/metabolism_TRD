# MwC 03 — Healthy-reference MwC deviations

This notebook is one ordered step in the analysis workflow.

```python
from pathlib import Path
import sys
import numpy as np
import pandas as pd
from IPython.display import Image, display

ROOT = Path.cwd()
if not (ROOT / 'analysis').is_dir():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'analysis'))
pd.set_option('display.max_columns', 30)
pd.set_option('display.max_rows', 100)
```

```python
from core import load_data
data = load_data(); meta = data['meta']; baseline = meta.session.eq('ses-01')
rows=[]
for cohort in ['HC','nTRD','TRD1','TRD2']:
    values=data['mwc_z'][baseline & meta.analysis_cohort.eq(cohort)]
    rows.append({'cohort':cohort,'participants':len(values),'finite_values':int(np.isfinite(values).sum()),
                 'mean_deviation':float(np.nanmean(values)),'SD':float(np.nanstd(values,ddof=1))})
display(pd.DataFrame(rows))
```
