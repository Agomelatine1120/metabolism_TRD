This notebook is one ordered step in the analysis workflow.

```python
from pathlib import Path
import sys
import numpy as np
import pandas as pd
from IPython.display import Image, display

ROOT = Path.cwd()
if not (ROOT / 'analysis').is_dir():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'analysis'))
pd.set_option('display.max_columns', 30)
pd.set_option('display.max_rows', 100)
```

```python
from core import load_data, longitudinal_results
tests=longitudinal_results(load_data())
mwc=tests.loc[tests.modality.eq('MwC')]
display(mwc.groupby('analysis_arm').agg(min_N=('n','min'),max_N=('n','max'),
                                        paired_p_lt_0p05=('paired_p_lt_0p05','sum'),
                                        paired_p_lt_0p01=('paired_p_lt_0p01','sum')).reset_index())
display(mwc.loc[mwc.paired_p_lt_0p05])
```
