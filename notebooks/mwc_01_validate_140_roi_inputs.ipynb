# MwC 01 — Validate 140-region inputs

This notebook is one ordered step in the analysis workflow.

```python
from pathlib import Path
import sys
import numpy as np
import pandas as pd
from IPython.display import Image, display

ROOT = Path.cwd()
if not (ROOT / 'analysis').is_dir():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'analysis'))
pd.set_option('display.max_columns', 30)
pd.set_option('display.max_rows', 100)
```

```python
import json
import numpy as np
with np.load(ROOT/'inputs'/'session_pet_gcmi_AAL3_140.npz', allow_pickle=False) as z:
    input_summary = {'sessions': len(z['participant_id']), 'nodes': len(z['roi_names']),
                     'PET_shape': z['pet_suvr'].shape, 'GCMI_shape': z['gcmi_bits'].shape,
                     'PET_nonfinite_values': int((~np.isfinite(z['pet_suvr'])).sum())}
with np.load(ROOT/'inputs'/'normative_SC_top10_mask.npz', allow_pickle=False) as z:
    input_summary['SC_shape'] = z['sc_top10_mask'].shape
    input_summary['SC_undirected_edges'] = int(np.count_nonzero(np.triu(z['sc_top10_mask'],1)))
contract = json.loads((ROOT/'inputs'/'analysis_contract.json').read_text())
display(pd.Series(input_summary, name='value').to_frame())
display(pd.DataFrame({'excluded_AAL3_region': contract['excluded_aal3_regions']}))
```
